```python
import numpy as np
import pandas as pd
import your
from tqdm import tqdm
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import sys
import matplotlib as mpl
from matplotlib import colors
from scipy.optimize import curve_fit, leastsq
import numpy.ma as ma
import matplotlib.ticker as ticker
from matplotlib.ticker import MaxNLocator

from typing import Union, List

import frb_functions
```

```python
def waterfaller_maker(burst):
    """
    ax1 ax3 ax5
    ax2 ax4 ax6

    the burst has to be a filpath and bursts are processed with YOUR
    """
    
    #Define the figure
    fig5 = plt.figure(figsize=(20,10))
    
    #Width 0.1 is for the hidden axes
    widths = [1, 1, 1]
    heights = [1, 2.5]
    spec5 = fig5.add_gridspec(ncols=3, nrows=2, width_ratios=widths, height_ratios=heights,\
                              wspace=0.25, hspace=0.0)
    
    ###################################
    ## Torun ## 
    burst = your.Your(burst_list[0])
    
    #Getting the info of the headerfile
    header_dict = frb_functions.your_header_to_dict(burst)
    tsamp = header_dict["tsamp"]
    nbins = header_dict["nspectra"]
    fsamp = header_dict["foff"] * -1
    freqs = header_dict["freqs"] # in Mhz
    
    #Extracting the extent
    tmin = 0.0
    tmax = nbins * burst.tsamp # seconds
    fmin = header_dict["fmin"]
    fmax = header_dict["fmax"]
    extent_b = [tmin, tmax, fmin, fmax]
    
    offset = 0.846
    extent_b[0] = (tmin - offset) * 1000
    extent_b[1] = (tmax - offset) * 1000
    print(f"the extent of the data is: {extent_b}")
    
    #time increasing, freq decreasing
    print("loading in the data")
    d = frb_functions.get_data(burst,nstart=0, nsamp=nbins, npoln=1)
    
    #Dedispersing the data
    dm_dedisp = 527.8
    print(f"de-dispersing to {dm_dedisp} pc/cc")
    stokes_i = frb_functions.dedisperse(d,dm_dedisp,freqs,tsamp)

    #Correcting the bandpass
    print(f"Correcting for the bandpass")
    stokes_i_norm = frb_functions.correct_bandpass(stokes_i, scale=3, sides="both")
    
    new_quantiles = np.percentile(stokes_i_norm, (1, 99.9))

    #Create x grid to plot the step function of the time series
    x_grid = np.linspace(extent_b[0], extent_b[1], endpoint=False, num=stokes_i_norm.shape[1])
    prof = np.mean(stokes_i_norm,axis=0)
    stokes_i_normprof = frb_functions.normalize_1d(prof,sides="both", scale=3)
    
    ax2 = fig5.add_subplot(spec5[1, 0])
    ax2.imshow(stokes_i_norm, aspect='auto', interpolation='None',\
          extent=extent_b, vmin=new_quantiles[0], vmax=new_quantiles[1])
    ax2.set_xlim(-15,15)
    ax2.set_ylabel('Frequency (MHz)',size=25)
    ax2.set_xlabel('Time (ms)',size=25)
    ax2.yaxis.set_major_locator(ticker.LinearLocator(5))
    ax2.xaxis.set_minor_locator(ticker.MultipleLocator(2.5))
    
    ax1 = fig5.add_subplot(spec5[0, 0], sharex=ax2)
    ax1.set_ylabel('S/N',size=25)
    ax1.plot(x_grid, stokes_i_normprof, drawstyle='steps-post', color='black')

    ax1.text(0.96, 0.93, "Toruń", transform=ax1.transAxes, fontsize=20, va='top', ha='right')
    #ax1.text(-28,12, f"DM = 219.37" + r" pc cm$^{-3}$", fontsize=18)
    
    #Ticks on the SN-
    maxSN = np.ceil(np.max(stokes_i_normprof))
    yticks_sn=np.array([0,int(maxSN/2),int(maxSN)])
    ax1.set_yticks(yticks_sn)
    ax1.set_ylim(-13)
    
    #ax1.yaxis.set_minor_locator(ticker.LinearLocator(3))
    #ax1.set_ylim(-5,maxSN+1)
    ax1.text(0.03, 0.93, fr"$\Delta$t = {(tsamp*1e6):.0f} $\mu$s" + "\n" + fr"$\Delta \nu$ = {(fsamp*1e3):.0f} kHz",\
             transform=ax1.transAxes, fontsize=18, va='top', ha='left')
    # ax1.text(0.7, 0.93, fr"{(tsamp*1e6):.0f} $\mu$s", transform=ax1.transAxes, fontsize=18, va='top', ha='left')
    # ax1.text(0.7, 0.78, fr"{(fsamp*1e3):.0f} kHz", transform=ax1.transAxes, fontsize=18, va='top', ha='left')
    
    ###################################

    ###################################
    # ## Westerbork - 1 ## 
    burst = your.Your(burst_list[1])
    
    #Getting the info of the headerfile
    header_dict = frb_functions.your_header_to_dict(burst)
    tsamp = header_dict["tsamp"]
    nbins = header_dict["nspectra"]
    fsamp = header_dict["foff"] * -1
    freqs = header_dict["freqs"] # in Mhz
    
    #Extracting the extent
    tmin = 0.0
    tmax = nbins * burst.tsamp # seconds
    fmin = header_dict["fmin"]
    fmax = header_dict["fmax"]
    extent_b = [tmin, tmax, fmin, fmax]
    
    offset = 1.765
    extent_b[0] = (tmin - offset) * 1000
    extent_b[1] = (tmax - offset) * 1000
    print(f"the extent of the data is: {extent_b}")
    
    #time increasing, freq decreasing
    print("loading in the data")
    d = frb_functions.get_data(burst,nstart=0, nsamp=nbins, npoln=1)

    #Dedispersing the data
    dm_dedisp = 527.8
    print(f"de-dispersing to {dm_dedisp} pc/cc")
    stokes_i = frb_functions.dedisperse(d,dm_dedisp,freqs,tsamp)

    #Correcting the bandpass
    print(f"Correcting for the bandpass")
    stokes_i_norm = frb_functions.correct_bandpass(stokes_i, scale=3, sides="both")

    downsamp_t = 16
    downsamp_f = 16
    print(f"Downsampling")
    stokes_i_norm = frb_functions.downsample(stokes_i_norm, freqf=downsamp_f, timef=downsamp_t)
    tsamp = tsamp * downsamp_t
    fsamp = fsamp * downsamp_f
    print(f"Correcting for the bandpass prt2")
    stokes_i_norm = frb_functions.correct_bandpass(stokes_i_norm, scale=3, sides="both")

    new_quantiles = np.percentile(stokes_i_norm, (1, 99.9))

    #Create x grid to plot the step function of the time series
    x_grid = np.linspace(extent_b[0], extent_b[1], endpoint=False, num=stokes_i_norm.shape[1])
    prof = np.mean(stokes_i_norm,axis=0)
    stokes_i_normprof = frb_functions.normalize_1d(prof,sides="both", scale=3)
    
    ax4 = fig5.add_subplot(spec5[1, 1])
    ax4.imshow(stokes_i_norm, aspect='auto', interpolation='None',\
          extent=extent_b, vmin=new_quantiles[0], vmax=new_quantiles[1])
    ax4.set_xlim(-100,100)
    ax4.set_xlabel('Time (ms)',size=25)
    ax4.yaxis.set_major_locator(ticker.LinearLocator(5))
    ax4.xaxis.set_minor_locator(ticker.MultipleLocator(25))

    ax3 = fig5.add_subplot(spec5[0, 1], sharex=ax4)
    ax3.set_ylabel('S/N',size=25)
    ax3.plot(x_grid, stokes_i_normprof, drawstyle='steps-post', color='black')

    ax3.text(0.96, 0.93, "Westerbork" + "\n" + "RT-1", transform=ax3.transAxes, fontsize=20, va='top', ha='right')
    #ax1.text(-28,12, f"DM = 219.37" + r" pc cm$^{-3}$", fontsize=18)
    
    #Ticks on the SN-
    maxSN = np.ceil(np.max(stokes_i_normprof))
    yticks_sn=np.array([0,int(maxSN/2),int(maxSN)])
    ax3.set_yticks(yticks_sn)
    
    #ax1.yaxis.set_minor_locator(ticker.LinearLocator(3))
    #ax1.set_ylim(-5,maxSN+1)
    ax3.text(0.03, 0.93, fr"$\Delta$t = {(tsamp*1e6):.0f} $\mu$s" + "\n" + fr"$\Delta \nu$ = {(fsamp*1e3):.0f} kHz",\
             transform=ax3.transAxes, fontsize=18, va='top', ha='left')
    # ax3.text(0.7, 0.93, fr"{(tsamp*1e6):.0f} $\mu$s", transform=ax3.transAxes, fontsize=18, va='top', ha='left')
    # ax3.text(0.7, 0.78, fr"{(fsamp*1e3):.0f} kHz", transform=ax3.transAxes, fontsize=18, va='top', ha='left')

    # ###################################

    # ###################################
    # ## Westerbork 2 ## 
    burst = your.Your(burst_list[2])
    
    #Getting the info of the headerfile
    header_dict = frb_functions.your_header_to_dict(burst)
    tsamp = header_dict["tsamp"]
    nbins = header_dict["nspectra"]
    fsamp = header_dict["foff"] * -1
    freqs = header_dict["freqs"] # in Mhz
    
    #Extracting the extent
    tmin = 0.0
    tmax = nbins * burst.tsamp # seconds
    fmin = header_dict["fmin"]
    fmax = header_dict["fmax"]
    extent_b = [tmin, tmax, fmin, fmax]
    
    offset = 1.551
    extent_b[0] = (tmin - offset) * 1000
    extent_b[1] = (tmax - offset) * 1000
    print(f"the extent of the data is: {extent_b}")
    
    #time increasing, freq decreasing
    print("loading in the data")
    d = frb_functions.get_data(burst,nstart=0, nsamp=nbins, npoln=1)

    #Dedispersing the data
    dm_dedisp = 527.8
    print(f"de-dispersing to {dm_dedisp} pc/cc")
    stokes_i = frb_functions.dedisperse(d,dm_dedisp,freqs,tsamp)

    #Correcting the bandpass
    print(f"Correcting for the bandpass")
    stokes_i_norm = frb_functions.correct_bandpass(stokes_i, scale=3, sides="both")

    downsamp_t = 2
    downsamp_f = 16
    print(f"Downsampling")
    stokes_i_norm = frb_functions.downsample(stokes_i_norm, freqf=downsamp_f, timef=downsamp_t)
    tsamp = tsamp * downsamp_t
    fsamp = fsamp * downsamp_f
    print(f"Correcting for the bandpass prt2")
    stokes_i_norm = frb_functions.correct_bandpass(stokes_i_norm, scale=3, sides="both")

    new_quantiles = np.percentile(stokes_i_norm, (1, 99.9))

    #Create x grid to plot the step function of the time series
    x_grid = np.linspace(extent_b[0], extent_b[1], endpoint=False, num=stokes_i_norm.shape[1])
    prof = np.mean(stokes_i_norm,axis=0)
    stokes_i_normprof = frb_functions.normalize_1d(prof,sides="both", scale=3)
    
    ax6 = fig5.add_subplot(spec5[1, 2])
    ax6.imshow(stokes_i_norm, aspect='auto', interpolation='None',\
          extent=extent_b, vmin=new_quantiles[0], vmax=new_quantiles[1])
    ax6.set_xlim(-20,20)
    ax6.set_xlabel('Time (ms)',size=25)
    ax6.yaxis.set_major_locator(ticker.LinearLocator(5))
    ax6.xaxis.set_minor_locator(ticker.MultipleLocator(5))

    ax5 = fig5.add_subplot(spec5[0, 2], sharex=ax6)
    ax5.set_ylabel('S/N',size=25)
    ax5.plot(x_grid, stokes_i_normprof, drawstyle='steps-post', color='black')

    ax5.text(0.96, 0.93, "Westerbork" + "\n" + "RT-1", transform=ax5.transAxes, fontsize=20, va='top', ha='right')
    #ax1.text(-28,12, f"DM = 219.37" + r" pc cm$^{-3}$", fontsize=18)
    
    #Ticks on the SN-
    maxSN = np.ceil(np.max(stokes_i_normprof))
    yticks_sn=np.array([0,int(maxSN/2),int(maxSN)])
    ax5.set_yticks(yticks_sn)
    
    #ax1.yaxis.set_minor_locator(ticker.LinearLocator(3))
    #ax1.set_ylim(-5,maxSN+1)
    ax5.text(0.03, 0.93, fr"$\Delta$t = {(tsamp*1e6):.0f} $\mu$s" + "\n" + fr"$\Delta \nu$ = {(fsamp*1e3):.0f} kHz",\
             transform=ax5.transAxes, fontsize=18, va='top', ha='left')
    # ax5.text(0.04, 0.78, , transform=ax5.transAxes, fontsize=18, va='top', ha='left')
    
    ###################################
    
    #Setting labels and ticks 
    for ax in [ax1,ax2,ax3,ax4,ax5,ax6]:
        
        ax.tick_params(labelsize=22) 
        
        for axis in ['top','bottom','left','right']:
            ax.spines[axis].set_linewidth(2)

            ax.tick_params(axis='y', which='major', direction='out', length=10, width=2)
            ax.tick_params(axis='y', which='minor', direction='out', length=5, width=2)
            ax.tick_params(axis='x', which='major', direction='out', length=10, width=2)
            ax.tick_params(axis='x', which='minor', direction='out', length=5, width=2)
    
    for ax in [ax1, ax3, ax5]:
        
        ax.tick_params(labelbottom=False)
        ax.tick_params(axis='x', which='major', direction='in', length=10, width=2)
        ax.tick_params(axis='x', which='minor', direction='in', length=5, width=2)
            
    plot_name = "atel_reactivation_frb240114"
    fig5.savefig(plot_name + ".pdf", dpi=600, bbox_inches='tight')
    
    fig5.savefig(plot_name + ".png", facecolor='white', \
            transparent=False, format = 'png',bbox_inches='tight', dpi=600)
    return
```
